```python
import pandas as pd

# Dữ liệu mô phỏng giao dịch thực tế
data = {
    "Ma_GD": ["GD01", "GD02", "GD03", "GD04", "GD05", "GD06", "GD07", "GD08"],
    "Phong_ban": ["Kế toán", "Tín dụng", "Kế toán", "Khách hàng DN", "Tín dụng", "Kế toán", "Tín dụng", "Khách hàng DN"],
    "Trang_thai": ["Hoàn thành", "Chờ duyệt", "Hoàn thành", "Hoàn thành", "Từ chối", "Chờ duyệt", "Hoàn thành", "Hoàn thành"],
    "So_tien": [15_000_000, 120_000_000, 25_000_000, 80_000_000, 45_000_000, 10_000_000, 60_000_000, 150_000_000]
}

df = pd.DataFrame(data)
print("=== DỮ LIỆU BAN ĐẦU ===")
print(df)
```

```python
# Tính tổng số tiền theo từng phòng ban
tong_theo_phong = df.groupby("Phong_ban")["So_tien"].sum().reset_index()

print("\n--- 1. TỔNG TIỀN THEO TỪNG PHÒNG BAN ---")
print(tong_theo_phong)
```

```python
# Dùng từ điển (dict) trong .agg() để chỉ định rõ từng phép tính
bao_cao_chi_tiet = df.groupby("Phong_ban").agg(
    So_luong_GD=("Ma_GD", "count"),      # Đếm số lượng mã GD
    Tong_so_tien=("So_tien", "sum"),     # Tính tổng tiền
    Trung_binh_GD=("So_tien", "mean")    # Giá trị giao dịch trung bình
).reset_index()

print("\n--- 3. BÁO CÁO TỔNG HỢP ĐA CHỈ SỐ (.agg) ---")
print(bao_cao_chi_tiet)
```

Hết phần 3

```python
bang_pivot = pd.pivot_table(
    df,
    values="So_tien",         # Cột cần tính toán giá trị
    index="Phong_ban",        # Cột đặt ở trục dòng (Rows)
    columns="Trang_thai",     # Cột đặt ở trục cột (Columns)
    aggfunc="sum",            # Phép tính: sum, count, mean... (mặc định là mean)
    fill_value=0,             # Điền số 0 vào các ô không có dữ liệu (tránh NaN)
    margins=True,             # Thêm dòng và cột Tổng cộng (Grand Total)
    margins_name="Tổng cộng"  # Tên hiển thị của dòng/cột tổng
)

print("\n--- 4. BẢNG MA TRẬN PIVOT TABLE HAI CHIỀU ---")
print(bang_pivot)
```

```python
with pd.ExcelWriter("bao_cao_tong_hop.xlsx", engine="openpyxl") as writer:
    bao_cao_chi_tiet.to_excel(writer, sheet_name="Chi_tiet_phong_ban", index=False)
    bang_pivot.to_excel(writer, sheet_name="Ma_tran_trang_thai")

print("\nĐã xuất báo cáo thành công ra file: bao_cao_tong_hop.xlsx")
```
